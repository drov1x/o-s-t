# 前情提要
该文件夹为数据结构笔记，考虑到数据结构需要基于C语言，因此使用cling作为jupyter内核
## 安装
首先需要安装conda
#### 创建conda环境
```shell
conda create -n cling
conda activate cling
```
切记不要指定python，避免冲突
#### 安装xeus-cling内核与jupyter
```shell
conda install jupyter xeus-cling -c conda-forge
```
#### 验证安装
```shell
jupyter kernelspec list
```
检查是否有xcpp开头的内核
#### 刷新环境
ctrl+shift+p >reload window，刷新vsc

刷新完选择内核即可
